**D2 : Student Marks**

```python
marks = [72, 85, 90, 65, 88, 76, 95, 58]

marks.sort(reverse=True)
print(marks)

print("Top 3:", marks[:3])
print("Bottom 3:", marks[-3:])

total = sum(marks)
average = total/len(marks)
print("Total:", total)
print("average:", average)
```

**D3 : Monthly Sales Report (Data Analytics style)**

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [12000, 15000, 9000, 18000, 21000, 17000]
print("Total:",sum(sales)) #sum()= Total
Average = sum(sales) /len(sales) # len - Function to count the length.
print("average:",round(Average,2)) # round - To round the value in fixed decimal place.
```

```python
high_sale = max(sales)
print(high_sale)
print("Best Month:",months[sales.index(high_sale)],"(",high_sale,")")
count = 0
for sale in sales:
    if sale >= Average:
        count = count+1
print("Month Above Average:",count)
```
