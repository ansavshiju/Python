*Question 2 : remove() vs discard()*

```python
colors = {"Red", "Green", "Blue"}

# a) Remove Green
colors.remove("Green")
print(colors)
```

```python
# b) Remove Yellow using remove()
colors.remove("Yellow")
```

*Question 3 : Set Operations*

```python
python_students = {"Arun", "Meera", "Rahul", "Sneha", "Kiran"}

sql_students = {"Meera", "Kiran", "Anil", "Divya"}
```

*Question 5 — Set vs Frozenset*

```python
# a) Normal set inside another set

s = {1, 2, 3}
print({s, 4})

# Why? - Normal set is mutable.
```

```python
# b) Frozenset inside another set

s = frozenset({1, 2, 3})
print({s, 4})
```

```python
# c) Explain why -
A normal set is mutable, so it cannot be used as a dictionary key or as an element inside another set.
A frozenset is immutable, so it can be used as a dictionary key or inside another set.
```

```python
# d)Real-life use case
# Set:Store unique student IDs.
student_ids = {101, 102, 103}
# Frozenset:Store a fixed group of days/categories that should not change.
days = frozenset(["Mon", "Tue", "Wed"])
```

*Question 7 — Three-Day Workshop*

```python
day1 = {"Anu", "Ben", "Cara", "Dev", "Esha", "Farhan"}

day2 = {"Ben", "Dev", "Esha", "Gita", "Hari"}

day3 = {"Dev", "Esha", "Hari", "Isha", "Anu"}
```

```python
#a) Attended all three days

answer = day1.intersection(day2, day3)
print(answer)
```

```python
# b) Attended at least one day

answer = day1.union(day2, day3)
print(answer)
```

```python
# c) Day 1 but missed Day 2 and Day 3

answer = day1 - day2 - day3
print(answer)
```

```python
# d) Day 1 and Day 2 but NOT Day 3

answer = day1.intersection(day2).difference(day3)
print(answer)
```

```python
# e) Exactly one day

day1_only = day1 - day2 - day3
day2_only = day2 - day1 - day3
day3_only = day3 - day1 - day2

answer = day1_only | day2_only | day3_only
print(answer)
```

*Question 8 — Unique Pairs Using Frozenset*

```python
pairs = [
    ("Asha", "Bala"),
    ("Bala", "Asha"),
    ("Chitra", "Dinesh"),
    ("Asha", "Chitra"),
    ("Dinesh", "Chitra"),
    ("Asha", "Bala")
]
```

```python
# a) Tuples directly into Set

unique_pairs = set(pairs)
print(unique_pairs)
print(len(unique_pairs))
```

```python
a) 5 unique pairs. That is wrong, because tuples are ordered, so ("Asha","Bala") and ("Bala","Asha") count as different. The correct answer is 3.
```

```python
b)frozenset(("Asha", "Bala")) 
The three conversations are {Asha, Bala}, {Chitra, Dinesh}, {Asha, Chitra}.
```

c) A normal set is mutable and therefore unhashable. Set elements and dict keys must be hashable, and frozenset is immutable and hashable.

```python
# d) Dictionary to count conversations

pairs = [
    ("Asha", "Bala"),
    ("Bala", "Asha"),
    ("Chitra", "Dinesh"),
    ("Asha", "Chitra"),
    ("Dinesh", "Chitra"),
    ("Asha", "Bala")
]

conversation_count = {}

for pair in pairs:
    conversation = frozenset(pair)

    if conversation in conversation_count:
        conversation_count[conversation] += 1
    else:
        conversation_count[conversation] = 1

print(conversation_count)
```
