```python
# 1. FizzBuzz
n = int(input("Number: "))
if n % 3 == 0 and n % 5 == 0:
    print("FizzBuzz")
elif n % 3 == 0:
    print("Fizz")
elif n % 5 == 0:
    print("Buzz")
else:
    print(n)
```

```python
# 3. BMI
w = float(input("Weight (kg): "))
h = float(input("Height (m): "))
bmi = w / h ** 2
print("BMI:", round(bmi, 2))
if bmi < 18.5:
    print("Underweight")
elif bmi < 25:
    print("Normal")
elif bmi < 30:
    print("Overweight")
else:
    print("Obese")
```

```python
# 6. Profit / Loss
cp = float(input("Cost price: "))
sp = float(input("Selling price: "))
if sp > cp:
    print("Profit:", sp - cp, "Percentage:", (sp - cp) / cp * 100, "%")
elif sp < cp:
    print("Loss:", cp - sp, "Percentage:", (cp - sp) / cp * 100, "%")
else:
    print("No profit no loss")
```

```python
#15. Why 0.1 + 0.2 == 0.3 is False
a = 0.1 + 0.2
b = 0.3

print(a == b)

if abs(a - b) < 0.000001:
    print("Equal")
else:
    print("Not Equal")
```

- 0.1 + 0.2 is slightly different from 0.3 because Python cannot store some decimal numbers exactly.
- Therefore, 0.1 + 0.2 == 0.3 gives False.
- abs(a - b) finds the small difference between the two values.
- If this difference is smaller than a small value like 0.000001, we can consider them equal.
- So, using abs() is a correct way to compare floating-point numbers
