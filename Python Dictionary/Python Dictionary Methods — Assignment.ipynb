*2. Ask the user for a key to delete from a dictionary. Use pop() with a default value so the program prints
either the removed value or &quot;Key not found&quot;. It must never crash.*


```python
student = {
    "name": "Rahul",
    "age": 21,
    "course": "Python"
}

key = input("Enter key to delete: ")

removed = student.pop(key, "Key not found")

if removed == "Key not found":
    print("Key not found")
else:
    print("Removed value:", removed)

print(student)
```

*5. Given marks = {&quot;Maths&quot;: 80, &quot;Science&quot;: 75, &quot;English&quot;: 90}, use a while loop and popitem() to remove and
print items one by one until the dictionary is empty.*



```python
marks = {
    "Maths": 80,
    "Science": 75,
    "English": 90
}
while marks:
    item = marks.popitem()
    print(item)
```

*6. Make a cart dictionary with 3 items. Take a backup using copy(), then clear() the cart. Print both and print
len() of each.*

```python
cart = {
    "pen": 10,
    "book": 50,
    "bag": 500
}

backup = cart.copy()

cart.clear()

print("Cart:", cart)
print("Backup:", backup)

print("Cart length:", len(cart))
print("Backup length:", len(backup))
```

**5. Shallow copy puzzle**

```python
a = {"name": "John", "marks": [80, 90]}
b = a.copy()
b["marks"].append(100)
b[ "name"] = "Sam"
print (a)
# {'name':"John""marks': [80, 90, 100]}
```

name is a string. b["name"] = "Sam" rebinds b's key to a new string, so a keeps "John". marks is a list shared by both dictionaries.
append(100) modifies that one list in place, so a sees the change.
