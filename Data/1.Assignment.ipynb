**Coding Problems**

**B1 — Name Formatter**

```python
name = input("Enter your full name: ")

name = " ".join(name.split())

print(name.title())
```

**B2 — Email Validator**

```python
email = input("Enter your email: ")

if "@" in email and email.endswith(".com") and not email.startswith("@"):
    print("Valid")
else:
    print("Invalid")
```

**B3 — Username Extractor**

```python
email = input("Enter your email: ")

position = email.find("@")

username = email[:position]
domain = email[position + 1:]

print("Username:", username)
print("Domain:", domain)
```

**B4 — Palindrome Checker**

```python
word = input("Enter a word: ")

word = word.lower()

if word == word[::-1]:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

**B5 — Mask the Phone Number**

```python
phone = input("Enter your 10-digit phone number: ")

masked = phone[:2] + "*" * 6 + phone[-2:]

print(masked)
```

**B6 — Initials Generator**

```python
name = input("Enter full name: ")

parts = name.split()

initials = [parts[0][0], parts[1][0], parts[2][0]]

print(".".join(initials))
```

**8B7 — Swap First and Last Character**

```python
word = input("Enter a word: ")

result = word[-1] + word[1:-1] + word[0]

print(result)
```

**B8 — Data Cleaning: Course List**

```python
raw = " python ,SQL, power bi ,EXCEL, gcp "

courses = [course.strip().title() for course in raw.split(",")]

print(courses)

print(" | ".join(courses))
```

**SECTION C — Student ID Card Generator**

```python
data = " rahul kumar,data analytics,rahul.kumar@gmail.com,9876543210 "

data = data.strip()

parts = data.split(",")

name = parts[0].strip().title()
course = parts[1].strip().title()
email = parts[2].strip().lower()
phone = parts[3].strip()

at_position = email.find("@")
username = email[:at_position]

phone_masked = phone[:2] + "*" * 6 + phone[-2:]

name_parts = name.split()
initials = name_parts[0][0] + "." + name_parts[1][0]

course_code = course[:3].upper() + "-2026"

print("========== STUDENT ID CARD ==========")
print("Name :", name)
print("Course :", course)
print("Email :", email)
print("Username :", username)
print("Phone :", phone_masked)
print("Initials :", initials)
print("Course Code :", course_code)
print("=====================================")
```
